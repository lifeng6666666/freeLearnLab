## 1. 环境配置

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import matplotlib.font_manager as fm
import warnings
warnings.filterwarnings('ignore')

# 中文字体配置
import os, platform
if platform.system() == 'Linux' and os.path.exists('/usr/share/fonts/truetype/chinese/NotoSansSC-Regular.ttf'):
    fm.fontManager.addfont('/usr/share/fonts/truetype/chinese/NotoSansSC-Regular.ttf')
plt.rcParams['font.sans-serif'] = ['Noto Sans SC', 'DejaVu Sans', 'Arial Unicode MS']
plt.rcParams['axes.unicode_minus'] = False
plt.rcParams['font.size'] = 11

# 统一配色
COLORS = {
    'primary': '#6b6144', 'accent': '#96781b', 'accent2': '#459ebb',
    'pos': '#c0392b', 'neg': '#2980b9', 'green': '#3d8e58',
    'orange': '#937b4a', 'muted': '#88867f'
}

print('环境配置完成 ✓')
```

## 2. 绘图工具函数

```python
def plot_confusion_matrix(cm, classes, title='混淆矩阵', cmap='Blues'):
    """绘制混淆矩阵热力图"""
    fig, ax = plt.subplots(figsize=(7, 6), constrained_layout=True)
    im = ax.imshow(cm, interpolation='nearest', cmap=cmap)
    ax.set_title(title, fontsize=13, fontweight='bold')
    fig.colorbar(im, ax=ax, shrink=0.8)
    ax.set_xticks(range(len(classes))); ax.set_yticks(range(len(classes)))
    ax.set_xticklabels(classes, rotation=45)
    ax.set_yticklabels(classes)
    thresh = cm.max() / 2
    for i in range(cm.shape[0]):
        for j in range(cm.shape[1]):
            ax.text(j, i, str(cm[i,j]), ha='center', va='center',
                    color='white' if cm[i,j] > thresh else 'black',
                    fontsize=10, fontweight='bold')
    ax.set_ylabel('真实标签'); ax.set_xlabel('预测标签')
    plt.show()


def plot_feature_importance(importances, names, title='特征重要性', top_n=10):
    """绘制特征重要性水平条形图"""
    indices = np.argsort(importances)[-top_n:]
    fig, ax = plt.subplots(figsize=(8, 5), constrained_layout=True)
    ax.barh(range(len(indices)), importances[indices], color=COLORS['accent'])
    ax.set_yticks(range(len(indices)))
    ax.set_yticklabels([names[i] for i in indices])
    ax.set_xlabel('重要性'); ax.set_title(title, fontsize=13, fontweight='bold')
    ax.grid(axis='x', alpha=0.3)
    plt.show()


def plot_learning_curve(train_scores, val_scores, title='学习曲线'):
    """绘制学习曲线"""
    fig, ax = plt.subplots(figsize=(8, 5), constrained_layout=True)
    ax.plot(train_scores, color=COLORS['accent'], lw=2, label='训练集')
    ax.plot(val_scores, color=COLORS['pos'], lw=2, label='验证集')
    ax.set_xlabel('迭代次数'); ax.set_ylabel('得分')
    ax.set_title(title, fontsize=13, fontweight='bold')
    ax.legend(); ax.grid(alpha=0.3)
    plt.show()

print('工具函数定义完成 ✓')
```
